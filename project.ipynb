# Manual neural-network project

Classification, regression, four activations and three gradient-descent strategies. All pictures, raw pixels and model checkpoints remain local.


```python
from pathlib import Path
import sys,json,numpy as np
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'network.py').exists())
sys.path.insert(0,str(ROOT))
from network import NeuralNetwork
```

## Mathematical check

A small separable fixture checks the implementation, not real-data generalization.


```python
x=np.array([[-2.,-1.],[-1.,-2.],[2.,1.],[1.,2.]])
y=np.array([0,0,1,1])
model=NeuralNetwork([2,6,2],activation='tanh')
model.fit(x,y,x,y,100,.1)
assert model.evaluate(x,y)[1]==1.0
```

## Experiment and fresh held-out results

Full retraining uses the documented command. Validation checkpoint selection excludes held-out test results.


```python
metrics=json.loads((ROOT/'metrics.json').read_text())
summary={group:{name:m['test'] for name,m in metrics[group]['models'].items()} for group in ['moons_classification','eggholder_regression','fashion_classification']}
assert all(m['checkpoint_reload_identical'] for group in ['moons_classification','eggholder_regression','fashion_classification'] for m in metrics[group]['models'].values())
summary
```
